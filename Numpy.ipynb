```python
import numpy as np
print("NumPy version:", np.__version__)
```

```python

# Load deliveries.csv
# dtype=str    → read everything as text first (mixed types in CSV)
# delimiter    → columns separated by comma
# skip_header  → skip row 0 (the column names row)

del_data = np.genfromtxt(
    'deliveries.csv',
    dtype=str,
    delimiter=',',
    skip_header=1
)

print("deliveries.csv shape:", del_data.shape)
print("First row preview:", del_data[0, :6])
```

```python

match_ids    = del_data[:, 0].astype(int)   
batting_team = del_data[:, 2]               
over         = del_data[:, 4].astype(int)   
ball         = del_data[:, 5].astype(int)   
batter       = del_data[:, 6]               
bowler       = del_data[:, 7]               
batsman_runs = del_data[:, 9].astype(int)   
total_runs   = del_data[:, 11].astype(int)  

print("Arrays extracted successfully!")
print(f"  match_ids    : {match_ids.shape}  → e.g. {match_ids[:3]}")
print(f"  batting_team : {batting_team.shape}  → e.g. {batting_team[:2]}")
print(f"  over         : {over.shape}  → e.g. {over[:5]}")
print(f"  batter       : {batter.shape}  → e.g. {batter[:2]}")
print(f"  bowler       : {bowler.shape}  → e.g. {bowler[:2]}")
print(f"  batsman_runs : {batsman_runs.shape}  → e.g. {batsman_runs[:5]}")
print(f"  total_runs   : {total_runs.shape}  → e.g. {total_runs[:5]}")
```

```python

# Load matches.csv  (needed for Task 10 — Toss Impact)

mat_data = np.genfromtxt(
    'matches.csv',
    dtype=str,
    delimiter=',',
    skip_header=1
)

m_match_ids   = mat_data[:, 0].astype(int)  
m_toss_winner = mat_data[:, 9]           

print("matches.csv shape:", mat_data.shape)
print(f"  m_match_ids   : {m_match_ids[:3]}")
print(f"  m_toss_winner : {m_toss_winner[:3]}")
```

 Task 1 — Total Runs per Match


```python
# Step 1: Get all unique match IDs (sorted automatically)
unique_matches = np.unique(match_ids)
print(f"Total unique matches: {len(unique_matches)}")
```

```python

match_totals = np.array([
    np.sum(batsman_runs[match_ids == mid])
    for mid in unique_matches
])


print(f"match_totals shape: {match_totals.shape}")
```

```python

print("Match ID      Total Runs")

for mid, total in zip(unique_matches[:10], match_totals[:10]):
    print(f"Match {mid}    {total} runs")
```

 Task 2 — Top 5 Batters by Total Runs


```python
# Step 1: All unique batter names (alphabetically sorted)
players = np.unique(batter)
print(f"Total unique batters: {len(players)}")
```

```python
# Total runs per player

player_runs = np.array([
    np.sum(batsman_runs[batter == p])
    for p in players
])
```

```python
#  argsort → get positions sorted ascending (lowest first)
sorted_idx = np.argsort(player_runs)

# Reverse for descending → slice top 5
top5_idx = sorted_idx[::-1][:5]

#  Use positions to get names and runs

print(f"{'Rank':<6} {'Batter':<25} {'Total Runs':>10}")

for rank, i in enumerate(top5_idx, 1):
    print(f"{rank:<6} {players[i]:<25} {player_runs[i]:>10}")
```

 Task 3 — Strike Rate of Each Batter


```python

players = np.unique(batter)

sr_vals = np.array([
    (np.sum(batsman_runs[batter == p]) / np.sum(batter == p)) * 100
    if np.sum(batter == p) > 0 else 0.0
    for p in players
])
```

```python
# Top 10 by Strike Rate
top10_sr_idx = np.argsort(sr_vals)[::-1][:10]

print(f"{'Rank':<6} {'Batter':<25} {'Strike Rate':>12}")

for rank, i in enumerate(top10_sr_idx, 1):
    balls_faced = np.sum(batter == players[i])
    print(f"{rank:<6} {players[i]:<25} {sr_vals[i]:>12.2f}  ({balls_faced} balls)")
```

 Task 4 — Economy Rate of Bowlers


```python

bowlers_list = np.unique(bowler)
```

```python


eco_vals = np.array([
    np.sum(total_runs[bowler == b]) / (np.sum(bowler == b) / 6)
    if np.sum(bowler == b) > 0 else 0.0
    for b in bowlers_list
])
```

```python
# Best 10 economy bowlers (minimum 10 overs bowled for fairness)
min_balls_mask = np.array([np.sum(bowler == b) for b in bowlers_list]) >= 60

filtered_bowlers = bowlers_list[min_balls_mask]
filtered_eco     = eco_vals[min_balls_mask]

best_eco_idx = np.argsort(filtered_eco)[:10]   # ascending = lowest economy first

print(f"{'Rank':<6} {'Bowler':<28} {'Economy Rate':>12}")

for rank, i in enumerate(best_eco_idx, 1):
    print(f"{rank:<6} {filtered_bowlers[i]:<28} {filtered_eco[i]:>12.2f}")
```

 Task 5 — Average Runs per Over (1 to 20)


```python

avg_per_over = np.array([
    np.mean(batsman_runs[over == o])
    for o in range(20)
])

print("Over   Avg Runs/Ball  Cricket Over")
print("-" * 38)
for i, avg in enumerate(avg_per_over):
    bar = '█' * int(avg * 8)
    print(f"  {i:2d}  |    {avg:.4f}     |  Over {i+1:2d}  {bar}")
```

 Task 6 — Boundary Analysis (Fours & Sixes)


```python
# np.sum(True/False array) counts the True values
total_fours = np.sum(batsman_runs == 4)
total_sixes = np.sum(batsman_runs == 6)
total_bounds = total_fours + total_sixes

print(f"Total Fours     : {total_fours}")
print(f"Total Sixes     : {total_sixes}")
print(f"Total Boundaries: {total_bounds}")
```

```python
# Which batting team hits the most boundaries?
boundary_mask = (batsman_runs == 4) | (batsman_runs == 6)

teams = np.unique(batting_team)

# For each team: count rows that are BOTH a boundary AND that team's row
team_boundaries = np.array([
    np.sum(boundary_mask & (batting_team == t))
    for t in teams
])

# Rank all teams
ranked_idx = np.argsort(team_boundaries)[::-1]


print(f"{'Rank':<6} {'Team':<35} {'Boundaries':>10}")

for rank, i in enumerate(ranked_idx, 1):
    fours_t = np.sum((batsman_runs == 4) & (batting_team == teams[i]))
    sixes_t = np.sum((batsman_runs == 6) & (batting_team == teams[i]))
    print(f"{rank:<6} {teams[i]:<35} {team_boundaries[i]:>6}  (4s:{fours_t} 6s:{sixes_t})")
```

 Task 7 — Death Overs Analysis (Overs 16–20)


```python
death_mask = (over >= 15) & (over <= 19)

total_death_runs = np.sum(batsman_runs[death_mask])
avg_death_runs   = np.mean(batsman_runs[death_mask])
total_death_balls = np.sum(death_mask)

print(f"Total balls in death overs : {total_death_balls}")
print(f"Total runs in death overs  : {total_death_runs}")
print(f"Average runs per ball      : {avg_death_runs:.4f}")
print(f"Boundaries in death overs  : {np.sum((batsman_runs[death_mask] == 4) | (batsman_runs[death_mask] == 6))}")
```

```python
# Which team scores most runs in death overs?
teams = np.unique(batting_team)

team_death_runs = np.array([
    np.sum(batsman_runs[death_mask & (batting_team == t)])
    for t in teams
])

ranked_idx = np.argsort(team_death_runs)[::-1]

print(f"{'Rank':<6} {'Team':<35} {'Death Runs':>10}")

for rank, i in enumerate(ranked_idx, 1):
    print(f"{rank:<6} {teams[i]:<35} {team_death_runs[i]:>10}")
```

Task 8 — Highest Scoring Match


```python
# We already have unique_matches and match_totals from Task 1
# match_totals[i] = total runs in unique_matches[i]

# np.argmax → position of the highest value
best_idx = np.argmax(match_totals)

best_match_id = unique_matches[best_idx]
best_runs     = match_totals[best_idx]

print(f"Highest Scoring Match")
print(f"  Match ID   : {best_match_id}")
print(f"  Total Runs : {best_runs}")
```

 Task 9 — Match Winner Approximation


```python
winners_list = []

for mid in unique_matches:
    # Mask for this match only
    match_mask = match_ids == mid

    # Get both teams that batted in this match
    teams_in_match = np.unique(batting_team[match_mask])

    if len(teams_in_match) < 2:
        continue   

    t1, t2 = teams_in_match[0], teams_in_match[1]

    # Runs for each team: match_mask AND team_mask combined
    r1 = np.sum(batsman_runs[match_mask & (batting_team == t1)])
    r2 = np.sum(batsman_runs[match_mask & (batting_team == t2)])

    winner = t1 if r1 > r2 else t2
    loser  = t2 if r1 > r2 else t1
    winners_list.append((mid, winner, max(r1,r2), loser, min(r1,r2)))

print(f"Total matches with result: {len(winners_list)}")
print("First 10 results:")
print(f"{'Match ID':<12} {'Winner':<35} {'Runs':>6}  vs  {'Runs':>6}  {'Loser'}")
for mid, w, wr, l, lr in winners_list[:10]:
    print(f"{mid:<12} {w:<35} {wr:>6}  vs  {lr:>6}  {l}")
```

 Task 10 — Toss Impact Analysis


```python
toss_scored_more = 0   # matches where toss winner scored more runs
toss_scored_less = 0   # matches where toss winner scored fewer runs

for mid in np.unique(match_ids):

    #  Step 1: Find toss winner from matches.csv
    
    row_idx = np.where(m_match_ids == mid)[0]

    if len(row_idx) == 0:
        continue   # this match ID not found in matches.csv

    toss_team = m_toss_winner[row_idx[0]]

    # Step 2: Get both teams in this match from deliveries
    match_mask = match_ids == mid
    teams_in_match = np.unique(batting_team[match_mask])

    if len(teams_in_match) < 2:
        continue

    # Step 3: Compute runs for toss winner 
    toss_runs = np.sum(batsman_runs[match_mask & (batting_team == toss_team)])

    #  Step 4: Get the other team's runs 
    other_teams = teams_in_match[teams_in_match != toss_team]
    if len(other_teams) == 0:
        continue

    other_runs = np.sum(batsman_runs[match_mask & (batting_team == other_teams[0])])

    # Step 5: Compare
    if toss_runs > other_runs:
        toss_scored_more += 1
    else:
        toss_scored_less += 1

total_matches = toss_scored_more + toss_scored_less
pct_more = (toss_scored_more / total_matches) * 100

print(f"Total matches analysed       : {total_matches}")
print(f"Toss winner scored MORE runs : {toss_scored_more} ({pct_more:.1f}%)")
print(f"Toss winner scored LESS runs : {toss_scored_less} ({100-pct_more:.1f}%)")
print()
if pct_more > 55:
    print(" Conclusion: Toss DOES give a scoring advantage")
elif pct_more < 45:
    print(" Conclusion: Toss HURTS scoring (teams choose to field)")
else:
    print("Conclusion: Toss has NO clear impact on runs scored (close to 50/50)")
```

Task 11 — Match Scorecard Generation


```python

for mid in unique_matches[:10]:

    match_mask     = match_ids == mid
    teams_in_match = np.unique(batting_team[match_mask])

    
    print(f"  MATCH ID: {mid}")
    

    scores = {}

    for t in teams_in_match:
        team_mask = match_mask & (batting_team == t)

        runs  = np.sum(batsman_runs[team_mask])
        fours = np.sum(batsman_runs[team_mask] == 4)
        sixes = np.sum(batsman_runs[team_mask] == 6)

        scores[t] = runs
        print(f"  {t:<40} {runs:>4} runs  (4s:{fours:>3}, 6s:{sixes:>3})")

    # Declare winner
    if len(teams_in_match) == 2:
        t1, t2 = teams_in_match[0], teams_in_match[1]
        winner = t1 if scores[t1] > scores[t2] else t2
        margin = abs(scores[t1] - scores[t2])
        print(f" ★  Winner: {winner}  (by {margin} runs)")

    print()
```
